## Handling Misssing Values 

```python
df["Gender"].dtypes
```

```python

print(df["Gender"].unique())
```

```python
print(df["Gender"].mode()[0])
df["Gender"] = df["Gender"].fillna(df["Gender"].mode()[0])
```

# Fix Inconsistancy

```python
df["Gender"] = df["Gender"].str.title()
print(df["Gender"].unique())
```

```python
print(df["Self_Employed"].isnull().sum())
```

```python
df["Self_Employed"] = df["Self_Employed"].fillna(df["Self_Employed"].mode()[0])
```

```python
print(df["Loan_Amount_Term"].isnull().sum())  
df["Loan_Amount_Term"] = df["Loan_Amount_Term"].fillna(df["Loan_Amount_Term"].median())  
```

```python
print(df["Credit_History"].isnull().sum())
df["Credit_History"] = df["Credit_History"].fillna(df["Credit_History"].mode()[0])  
```

# Fix Inconsistancy

```python
df["Credit_History"] = df["Credit_History"].astype(int) 
print(df["Credit_History"].unique())
```

```python
print(df["Dependents"].isnull().sum())
print(df["Dependents"].unique())
print(df["Dependents"].mode()[0])
df["Dependents"] = df["Dependents"].fillna(df["Dependents"].mode()[0])
```

# Fix Inconsistancy

```python
df["Dependents"] = df["Dependents"].astype(str).str.strip()
print(df["Dependents"].unique())
```

# Univariate analysis on categorical column

```python
plt.figure(figsize=(4,4))
sns.countplot(x="Credit_History", data=df, palette="muted")
plt.title("Distribution of Credit History")
plt.xlabel("Credit History (0 = Bad, 1 = Good)")
plt.ylabel("Count")
plt.show()
```

# Univariate Analysis on Numerical column

```python
sns.histplot(df["LoanAmount"], bins=30, kde=True, color="cyan")
plt.title("Distribution of Loan Amounts")
plt.xlabel("Loan Amount")
plt.ylabel("Frequency")
plt.show()
```

# Bivariate Analysis : Categorical and Categorical

```python
plt.figure(figsize=(6,4))
sns.countplot(x="Education", hue="Loan_Status", data=df, palette="muted")
plt.title("Education vs Loan Status")
plt.xlabel("Education")
plt.ylabel("Count")
plt.show()
```

# Bivariate Analysis : Categorical and Numerical

```python
# Average loan amount by Education
df.groupby("Education")["LoanAmount"].mean()
```

```python
sns.boxplot(x="Education", y="LoanAmount", data=df, palette="Set1") 
plt.title("Loan Amount by Education Level") 
plt.show()
```

# Bivariate Analysis : Categorical and Numerical


```python
print(df.groupby("Property_Area")["LoanAmount"].mean())
plt.figure(figsize=(6,3)) 
sns.barplot(x="Property_Area", y="LoanAmount", data=df, palette="Set2")
plt.title("Loan Amount by Property Area") 
plt.xlabel("Property Area") 
plt.ylabel("Loan Amount") 
plt.show()
```

# Multivariate Analysis

```python
ct = pd.crosstab([df["Gender"], df["Education"]], df["Loan_Status"])
print(ct)
```

```python
sns.heatmap(ct, annot=True, cmap="coolwarm", fmt="d") 
plt.title("Loan Status by Gender and Education")
plt.show()
```
